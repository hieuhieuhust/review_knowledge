```python
from async_pytorch_engine import searchrequest, asynpytorchengine
import asyncio
```

```python
u1 = searchrequest(
    user_id= "user1",
    query_text= "chính sách bảo hiểm y tế doanh nghiệp",
    top_k= 2
)
u2 = searchrequest(
    user_id= "user2",
    query_text="quy trình tạm ứng công tác phí",
    top_k= 2
)
u3 = searchrequest(
    user_id= "user3",
    query_text="cách cấu hình GPU CUDA cho pytorch",
    top_k=2
)
u4 = searchrequest(
    user_id= "user4",
    query_text="hướng dẫn nộp thuế thu nhập cá nhân",
    top_k=2
)

user = asynpytorchengine()
```

```python
async def main():
    topku1 = user.search_single_query(u1)
    topku2 = user.search_single_query(u2)
    topku3 = user.search_single_query(u3)
    topku4 = user.search_single_query(u4)
    results = await asyncio.gather(
        topku1,
        topku2,
        topku3,
        topku4
    )
    for result in results:
        print(result)
    text = f"user 1 khớp nhất với tài liệu số {results[0]}"
    async for token in user.stream_token(text):
        print(token, end="", flush=True)
await main()
```
