```python
from pytorch_engine import DocumentChunk, pytorchvectormatcher
from loguru import logger
import torch
```

khởi tạo 3 tài liệu giả lập với documentchunk

```python
doc1 = DocumentChunk(
    doc_id = "doc1",
    text= "quy định về thời gian làm việc từ xa",
    vector_dim= 768
)
doc2 = DocumentChunk(
    doc_id = "doc2",
    text= "chính sách thưởng hiệu suất",
    vector_dim= 768
)
doc3 = DocumentChunk(
    doc_id = "doc3",
    text= "quy trình nộp đơn xin nghỉ phép có lương",
    vector_dim= 768
)
```

tạo 3 tensor cho 3 tài liệu

```python
pvm = pytorchvectormatcher()
doc1_vec= pvm.create_fake_embedding(24)
doc2_vec= pvm.create_fake_embedding(24)
doc3_vec= pvm.create_fake_embedding(24)

print(doc1_vec[:,0:3] - doc2_vec[:,0:3])
```

giả lập câu hỏi người dùng

```python
query = DocumentChunk(
    doc_id = "query",
    text = "tôi muốn xin nghỉ 2 ngày tuần tới thì làm như nào",
    vector_dim=768
)
query_vec = pvm.create_fake_embedding(25)
```

tính score của 2 cái 

```python
score = pvm.compute_similarity(query_vec, doc3_vec)
print(score)
for char in pvm.stream_matched_result(query.text, score):
    print(char,end="", flush=True)

torch.save(doc3_vec, f"storage/doc3_embedding.pt")
logger.success(f"đã lưu thành công")
```
