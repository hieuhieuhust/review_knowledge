NOTE CÁCH DÙNG CÁI @

+ tới @ thì chạy bên trong hàm bên cạnh @, chưa chạy trong wrapper
+ tới khi gọi hàm bên dưới @ (gọi phải có tham số chứ không phải gọi lúc gặp ngay bên dưới @) kể cả có tham số hay không thì nó sẽ chạy hàm wrapper in ra kết quả
+ nếu trong wrapper mà còn gọi hàm bên dưới @ thì lúc này sẽ thực sự chạy cái hàm bên dưới cái @
+ cứ hễ gọi hàm bên dưới @ (không tính hàm bên trong wrapper) thì luôn luôn, luôn luôn chạy cái wrapper với tham số là tham số của hàm bên dưới @
