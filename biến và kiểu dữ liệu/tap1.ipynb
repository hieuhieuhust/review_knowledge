```python
from typing import List, Dict, Any
from transformers import AutoTokenizer

# kiểu dữ liệu 
model_name: str = "gpt-4o-mini"
temperature: float = 0.7
max_tokens: int = 4096
is_active: bool = True

print(f"đang dùng model {model_name}, nhiệt độ {temperature}")

# 1 list: danh sách có thứ tự
model: List = ["gpt-4o-mini", "llama3.3"]
model.append("claude opus 4.0")

# 2 dict: cặp key-value 
message: Dict[str, str] = {
    "role": "user",
    "content": "giải thích RAG là gì trong 1 câu"
}
chat_history: List[Dict[str, str]]= [
    {"role": "system" , "content": "bạn là 1 trợ lý AI tiếng việt hữu ích"},
    {"role": "user" , "content": "xin chào"},
    {"role": "assitant", "content": "chào bạn, tôi có thể giúp gì cho bạn hôm nay"}
]
# in ra trường
for i in chat_history:
    print(i['role'].upper())
    
def validate_promt(promt: str, max_chars: int = 500) -> bool:
    # xóa khoảng trắng thừa trong promt
    clean_promt = promt.strip()
    
    if len(promt) == 0:
        print("lỗi promt không được để trống")
        return False
    elif len(clean_promt) > max_chars:
        print(f"promt không được vượt quá {max_chars} ký tự")
        return False
    else:
        return True
user_input: str = "xin chào ///"
if validate_promt(user_input):
    print("promt hợp lệ gửi tới LLM")

tokenizer = AutoTokenizer.from_pretrained(
    "Qwen/Qwen2.5-3B-Instruct"
)
    
def calculate_token_cost(text: str) -> Dict[str,Any]:
    estimated_token =  len(tokenizer.encode(text))
    cost_usd = estimated_token * 0.15 /1000000
    return {"text": text, "estimated token": estimated_token, "cost":cost_usd}

calculate_token_cost(user_input)

def add_message(role: str, content: str, chat: List[Dict[str,str]]):
    d: Dict[str, str] = {}
    d.update({
        "role": role, 
        "content": content
        })
    chat.append(d)
    return chat

def print_chat_transcript(chat: List[Dict[str,str]]):
    for messagee in chat:
        role = messagee["role"].upper()
        content = messagee["content"]
        print(f"{role}> {content}")

add_message(
    "user",
    "xin chào",
    chat_history
)     
print_chat_transcript(chat_history)
```
